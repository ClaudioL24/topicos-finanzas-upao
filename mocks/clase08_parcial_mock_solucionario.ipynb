## Caso 1: costo de capital

**Textil Norte S.A.** estima su costo de capital con la siguiente información: tasa libre de riesgo 4.0%, prima por riesgo de mercado 5.0%, prima por riesgo país 2.0% (con $\lambda = 1$), beta del patrimonio 1.05, estructura de capital objetivo a valores de mercado de 35% deuda y 65% patrimonio, costo de la deuda antes de impuestos 6.8% y tasa de impuesto a la renta 29.5%.

```python
# Caso 1: Textil Norte S.A.
t = 0.295
ke = capm(rf=0.04, beta=1.05, erp=0.05, crp=0.02)
ke_r = round(ke, 4)                                   # el item 12 fija Ke redondeado a dos decimales en porcentaje
k = wacc(E=65, D=35, ke=ke_r, kd=0.068, t=t)
print(f"11. Ke = {ke:.2%}   (sin prima pais: {capm(0.04, 1.05, 0.05, 0):.2%})")
print(f"12. WACC = {k:.2%}   (sin escudo fiscal: {wacc(65, 35, ke_r, 0.068, 0):.2%})")
```

*Ampliación.* El CAPM ampliado con prima país (semana 2): el accionista exige la tasa libre de riesgo más una compensación por el riesgo sistemático ($\beta \times ERP$) más, en un mercado emergente, la prima por riesgo país con $\lambda = 1$ (se suma completa, sin multiplicarla por el beta). Cada insumo debe ser consistente en moneda y plazo con los flujos que se van a descontar.

*Referencia:* Semana 2, costo de capital, CAPM y WACC (`clase02_apuntes.ipynb`).

**12. Clave B.** WACC $= 0.65(11.25) + 0.35(6.8)(1 - 0.295) = 8.99\%$. Omitir el escudo fiscal arroja 9.69%; invertir los pesos arroja 7.05%.

*Ampliación.* El WACC pondera el costo de cada fuente por su peso a valor de mercado en la estructura objetivo. La deuda entra neta de impuestos, $K_d(1-t)$, porque los intereses son deducibles y reducen el impuesto que paga la empresa. Si se omite el escudo fiscal el WACC sube; si se invierten los pesos se le da a la deuda el peso del patrimonio. El WACC es la tasa para flujos de la firma (FCFF), no para flujos del accionista.

*Referencia:* Semana 2, costo de capital, CAPM y WACC (`clase02_apuntes.ipynb`).


## Caso 2: flujos de caja libres (FCFF y FCFE)

**Distribuidora Andina S.A.** reportó en su último año (millones de soles): utilidad neta 140, depreciación 45, gasto por intereses 25, inversión en activo fijo (capex) 80, aumento del capital de trabajo 10 y endeudamiento neto (deuda nueva menos amortizaciones) 20. La tasa de impuesto es 30%.

```python
# Caso 2: Distribuidora Andina S.A.
t = 0.3
fcff = fcff_desde_ni(ni=140.0, ncc=45.0, interes=25.0, t=t, fcinv=80.0, wcinv=10.0)
fcfe = fcfe_desde_fcff(round(fcff, 1), interes=25.0, t=t, endeudamiento_neto=20.0)
print(f"13. FCFF = {fcff:.1f}   (interes sin ajuste tributario: {fcff + 25.0 * t:.1f})")
print(f"14. FCFE = {fcfe:.1f}   (restando el interes bruto: {round(fcff, 1) - 25.0 + 20.0:.1f})")
```

*Referencia:* Semana 3, valoración DCF, FCFF y FCFE (`clase03_apuntes.ipynb`).

*Ampliación.* Del FCFF al FCFE se pasa restando lo que se lleva el acreedor, el interés neto Int$(1-t)$, y sumando el endeudamiento neto: la deuda nueva es caja que entra para el accionista y las amortizaciones son caja que sale. El FCFE se descuenta al $K_e$ y da directamente el valor del patrimonio; el FCFF se descuenta al WACC y da el valor de la firma. Mezclarlos es el error de consistencia flujo-tasa.

*Referencia:* Semana 3, valoración DCF, FCFF y FCFE (`clase03_apuntes.ipynb`).


## Caso 3: valor terminal y valoración por múltiplos

Un analista valora **Energía del Sur S.A.A.**. El FCFF proyectado para el año 3 es 72 (millones de soles) y desde el año 4 se espera un crecimiento perpetuo de 2%, con un WACC de 9.0%. ## Caso 4: riesgo y portafolio de dos activos

Una gestora de portafolios analiza la acción X bajo tres escenarios para el próximo año: auge (probabilidad 0.25, retorno 28%), normal (0.45, 12%) y recesión (0.30, $-15$%). La acción Y tiene retorno esperado de 6% y desviación estándar de 9%. La correlación entre X e Y es 0.30. La gestora evalúa un portafolio con 35% en X y 65% en Y.

```python
# Caso 4: escenarios y dos activos (en porcentaje)
e_x, s_x = estadisticos_escenarios([0.25, 0.45, 0.3], [28.0, 12.0, -15.0])
s_x_r = round(s_x, 1)                                 # el item 18 fija sigma_X redondeado a un decimal
e_p, s_p = portafolio_dos_activos(0.35, e_x, 6.0, s_x_r, 9.0, 0.3)
print(f"17. E(R_X) = {e_x:.2f}% | sigma_X = {s_x:.2f}%   (sin probabilidades: {np.std([28.0, 12.0, -15.0]):.2f}%)")
print(f"18. cov = {0.3 * s_x_r * 9.0:.2f} | sigma_p = {s_p:.2f}%   (promedio ponderado: {0.35 * s_x_r + 0.65 * 9.0:.2f}%)")
```

*Ampliación.* Con escenarios, el retorno esperado es el promedio ponderado por probabilidades y la varianza es el promedio ponderado de las desviaciones al cuadrado respecto de ese retorno esperado; la desviación estándar es su raíz. Las probabilidades deben usarse en las dos cuentas: omitirlas (tratar los tres escenarios como equiprobables) o promediar desviaciones absolutas en lugar de cuadrados son los errores típicos. Conviene observar qué escenario aporta más varianza: casi siempre la recesión.

*Referencia:* Semana 6, riesgo y rendimiento (`clase06_apuntes.ipynb`).

**18. Clave A.** $\sigma_{XY} = 0.30(16.3)(9) = 44.01$; $\sigma_p^2 = (0.35)^2(16.3)^2 + (0.65)^2(9)^2 + 2(0.35)(0.65)(44.01) = 86.79$; $\sigma_p = 9.32\%$. El promedio ponderado de volatilidades arroja 11.55% (válido únicamente con $\rho = 1$); omitir la covarianza arroja 8.17%.

*Ampliación.* La varianza de dos activos tiene tres términos: los dos propios y el cruzado, $2 w_X w_Y \sigma_{XY}$, con $\sigma_{XY} = \rho\,\sigma_X\sigma_Y$. Con $\rho < 1$ la volatilidad del portafolio es menor que el promedio ponderado de las volatilidades: esa diferencia es la ganancia por diversificación. Promediar volatilidades solo es válido con $\rho = 1$; omitir la covarianza equivale a suponer $\rho = 0$ y subestima el riesgo cuando la correlación es positiva.

*Referencia:* Semana 6, riesgo y rendimiento (`clase06_apuntes.ipynb`).


## Caso 5: asignación óptima de capital y cartera completa

Un asesor financiero identificó el portafolio tangente, con retorno esperado de 10% y desviación estándar de 13%. La tasa libre de riesgo es 3.5% y es posible prestar y endeudarse a dicha tasa. Su cliente tiene una función de utilidad $U = E(R) - \tfrac{1}{2}A\sigma^2$ con $A = 3$ (retornos en decimales).

```python
# Caso 5: cartera completa (en decimales)
e_t, s_t, rf, A = 0.1, 0.13, 0.035, 3
y = (e_t - rf) / (A * s_t ** 2)
y_r = round(y, 2)
e_c, s_c = rf + y_r * (e_t - rf), y_r * s_t
print(f"19. sigma_T^2 = {s_t ** 2:.4f} | y* = {y:.4f} ({y:.1%})   (sin el cuadrado: {(e_t - rf) / (A * s_t):.1%} | sin restar R_f: {e_t / (A * s_t ** 2):.1%})")
print(f"20. E(R_c) = {e_c:.2%} | sigma_c = {s_c:.2%}   (y* x E(R_T): {y_r * e_t:.2%})")
```

**19. Clave C.** $y^* = [E(R_T) - R_f]/(A\,\sigma_T^2) = (0.10 - 0.035)/(3 \times 0.13^2) = 1.2821$, es decir 128.2%. Omitir el cuadrado de $\sigma_T$ arroja 16.7%; omitir la resta de $R_f$ arroja 197.2%.

*Ampliación.* Una vez identificado el portafolio tangente, la única decisión personal es cuánto invertir en él y cuánto en el activo libre de riesgo (separación en dos fondos). Maximizar $U = E(R) - \tfrac{1}{2}A\sigma^2$ sobre la CAL da $y^* = [E(R_T) - R_f]/(A\,\sigma_T^2)$, con retornos en decimales: la exposición crece con la prima del tangente y cae con su varianza y con la aversión al riesgo. Un $y^*$ mayor que uno significa endeudarse a $R_f$ para invertir más del 100 por ciento en $T$.

*Referencia:* Semana 7, Markowitz, CAL y cartera completa (`clase07_apuntes.ipynb`).

**20. Clave C.** $E(R_c) = R_f + y^*[E(R_T) - R_f] = 3.5 + 1.28(10 - 3.5) = 11.82\%$ (y $\sigma_c = 1.28 \times 13 = 16.64\%$). Multiplicar $y^*$ por $E(R_T)$ arroja 12.80%; sumar $R_f$ íntegro más $y^* E(R_T)$ arroja 16.30%.

*Ampliación.* Sobre la CAL, la cartera completa tiene $E(R_c) = R_f + y\,[E(R_T) - R_f]$ y $\sigma_c = y\,\sigma_T$: el activo libre de riesgo aporta su tasa pero ninguna varianza. Los dos errores frecuentes son multiplicar $y$ por el retorno total de $T$ (olvidando que la parte en $R_f$ también rinde) o sumar $R_f$ íntegro más $y\,E(R_T)$ (contando dos veces). El ratio de Sharpe de la cartera completa es el mismo que el de $T$: moverse por la CAL cambia el riesgo, no la calidad de la recompensa por riesgo.
