## 1. La macroeconomía entra por el numerador y por el denominador

Todo activo vale el valor presente de sus flujos esperados:

$$P_0 = \sum_{t=1}^{N} \frac{E(CF_t)}{(1 + r_t)^t}$$

Las noticias macroeconómicas mueven el **numerador** (utilidades, dividendos, probabilidad de pago) y el **denominador** (tasa real, inflación esperada, primas por riesgo). El efecto sobre el precio depende de cuál domine; por eso una misma noticia puede subir el precio de los bonos y bajar el de las acciones.

## 2. Componentes de la tasa de descuento

| Activo | Tasa de descuento |
|---|---|
| Bono soberano indexado a la inflación | $l$ |
| Bono soberano nominal | $l + \theta + \pi$ |
| Bono corporativo | $l + \theta + \pi + \gamma$ |
| Acción | $l + \theta + \pi + \gamma + \kappa$ |

Todas las primas compensan lo mismo: el activo paga poco justo cuando el inversionista más valora el dinero, es decir, en las recesiones. El $R_f$ de nuestro CAPM es la segunda fila; la prima de mercado resume $\gamma$ y $\kappa$ para el conjunto de las acciones.

**Tasa real y crecimiento.** La tasa real es el precio de trasladar consumo del presente al futuro. Sube con el crecimiento esperado (si mañana serás más rico, ahorras menos hoy) y las primas por riesgo suben con la incertidumbre sobre ese crecimiento. La tasa neutral $r^*$ se aproxima con el crecimiento potencial. Consecuencia para la valorización: la tasa de largo plazo y el $g$ perpetuo no son independientes (semana 4).

## 3. Inflación: Fisher e inflación implícita

$$(1 + i) = (1 + r)(1 + \pi^e) \qquad\Rightarrow\qquad i \approx r + \pi^e$$

```python
nominal, indexado = 0.045, 0.020          # rendimientos a 10 anos
bei_exacta = (1 + nominal) / (1 + indexado) - 1
bei_aprox = nominal - indexado
print(f"Inflacion implicita exacta:     {bei_exacta:.2%}")
print(f"Inflacion implicita aproximada: {bei_aprox:.2%}")

tasa, inflacion_esperada, inflacion_observada = 0.0575, 0.030, 0.035
print(f"Tasa real ex ante: {tasa - inflacion_esperada:.2%}")
print(f"Tasa real ex post: {tasa - inflacion_observada:.2%}")
```

Salida esperada: inflación implícita de 2.45 por ciento (2.50 con la aproximación); tasa real ex ante de 2.75 por ciento y ex post de 2.25.

Dos precisiones. La inflación implícita contiene la inflación esperada **y** la prima por incertidumbre de inflación, así que sobreestima la expectativa cuando esa prima es positiva. Y las decisiones de inversión se toman con la tasa real **ex ante**, porque la inflación que importa es la que viene. En la PC2 se acepta la tasa real calculada con la inflación observada cuando no encuentres la expectativa; declara cuál usaste.

## 4. Fases del ciclo económico

| Fase | Actividad | Inflación | Política monetaria | Curva |
|---|---|---|---|---|
| Recuperación | brecha negativa que se cierra | baja | expansiva | empinada |
| Expansión | crecimiento sobre el potencial | al alza | retiro del estímulo | se aplana |
| Desaceleración | crecimiento pierde impulso | alta | contractiva | plana o invertida |
| Contracción | crecimiento negativo | cede | recortes de tasa | se empina |

Para identificar la fase se usan varios indicadores a la vez: crecimiento frente al potencial, inflación frente a la meta, desempleo y dirección reciente de la tasa de referencia. Los mercados anticipan: los precios cambian de tendencia antes que los indicadores de actividad. La tabla es una guía, no una regla.

## 5. Política monetaria: regla de Taylor y postura

$$i = r^* + \pi + 0.5\,(\pi - \pi^*) + 0.5\,(y - y^*)$$

```python
def taylor(r_neutral, inflacion, meta, brecha):
    return r_neutral + inflacion + 0.5 * (inflacion - meta) + 0.5 * brecha

def postura(tasa, inflacion_esperada, r_neutral, tolerancia=0.25):
    real = tasa - inflacion_esperada
    if real > r_neutral + tolerancia:
        return real, "contractiva"
    if real < r_neutral - tolerancia:
        return real, "expansiva"
    return real, "neutral"

i_taylor = taylor(2.0, 3.5, 2.0, -1.0)                      # todo en %
print(f"Tasa segun la regla de Taylor: {i_taylor:.2f}%")
real, lectura = postura(i_taylor, 3.0, 2.0)
print(f"Tasa real ex ante: {real:.2f}% frente a r* = 2.00% -> postura {lectura}")
print(f"Item 3: {taylor(1.5, 4.0, 3.0, 2.0):.2f}%")
```

Salida esperada: 5.75 por ciento según la regla; tasa real ex ante de 2.75, mayor que la neutral de 2.00: postura contractiva. El ítem 3 da 7.00.

La postura no se juzga por el nivel de la tasa nominal: una tasa de 7 por ciento es expansiva si la inflación esperada es 8. Se juzga comparando la tasa real ex ante con la neutral. La tolerancia de 0.25 puntos del código es una convención para no llamar contractiva a una diferencia de centésimas; $r^*$ no es observable y se estima con error.

## 6. El ciclo en la renta fija y en la renta variable

**Renta fija.** El tramo corto de la curva sigue a la tasa de referencia; el largo refleja crecimiento e inflación esperados más primas por plazo. Los diferenciales de crédito se amplían en la desaceleración y la contracción y se reducen en la recuperación. Los bonos soberanos suelen subir de precio en una recesión (la tasa cae), salvo que coincida con inflación alta.

**Renta variable.** En $P_0 = D_1/(K_e - g)$ el ciclo mueve los tres insumos: las utilidades son procíclicas y más volátiles que el PBI; la prima por riesgo es **anticíclica** (sube en las contracciones); el P/E cae al inicio de la contracción y se recupera antes que las utilidades. Sectores cíclicos (materiales, industria, consumo discrecional, finanzas) con betas mayores que uno; defensivos (consumo básico, servicios públicos, salud) con betas menores que uno. Nuestra minera ($\beta = 1.2$) es cíclica; la eléctrica ($\beta = 0.6$), defensiva.

```python
def gordon(d1, rf, beta, prima, g):
    ke = rf + beta * prima
    return ke, d1 / ((ke - g) / 100)

d1, beta = 2.00, 1.2
escenarios = {"Base": (4.5, 5.5, 5.0),
              "Recesion esperada": (4.5, 6.5, 4.0),
              "Recesion con recorte de tasa": (3.5, 6.5, 4.0)}
ke0, p_base = gordon(d1, 4.5, beta, 5.5, 5.0)
for nombre, (rf, prima, g) in escenarios.items():
    ke, p = gordon(d1, rf, beta, prima, g)
    print(f"{nombre:30s} Ke = {ke:5.1f}%  g = {g:.1f}%  P0 = {p:6.2f}  variacion = {p / p_base - 1:+.1%}")
```

Salida esperada: 32.79 en el escenario base; 24.10 con recesión esperada ($-26.5$ por ciento); 27.40 si además el banco central recorta un punto ($-16.4$ por ciento).

Lo que enseña el ejemplo: el precio cae 26.5 por ciento **sin que cambie el dividendo del próximo año**. Basta que la prima suba un punto y el crecimiento esperado baje un punto, porque el valor depende de la diferencia $K_e - g$, que pasa de 6.1 a 8.3 puntos. El recorte de tasa compensa solo una parte. Es la misma sensibilidad de la tabla de la semana 4.

## 7. Tipo de cambio y retorno en dólares

$$(1 + R_{USD}) = (1 + R_{local})\,(1 + \Delta s)$$

donde $\Delta s$ es la variación porcentual del valor de la moneda local frente al dólar (negativa si se deprecia).

```python
def retorno_usd(r_local, variacion_moneda):
    return (1 + r_local) * (1 + variacion_moneda) - 1

print(f"Indice +12% y moneda -5%: {retorno_usd(0.12, -0.05):.2%}")
print(f"Item 4, indice +8% y moneda +4%: {retorno_usd(0.08, 0.04):.2%}")

# Si el dato es el tipo de cambio en moneda local por dolar (soles por dolar):
tc_inicial, tc_final = 3.70, 3.90
variacion_moneda = tc_inicial / tc_final - 1
print(f"El tipo de cambio sube de {tc_inicial} a {tc_final}: el sol vale {variacion_moneda:+.2%} en dolares")
```

Salida esperada: 6.40 por ciento en el primer caso y 12.32 en el segundo. Con el tipo de cambio de 3.70 a 3.90, el sol pierde 5.13 por ciento de su valor en dólares.

Cuidado con la convención: si el tipo de cambio se expresa en moneda local por dólar, una **subida** del tipo de cambio es una **depreciación** de la moneda local, y $\Delta s = TC_0/TC_1 - 1$. Es el cálculo de la sección de mercados del informe país.

**Economía pequeña y abierta.** En el Perú, los términos de intercambio son una fuente principal del ciclo: un aumento del precio del cobre eleva utilidades mineras, ingreso fiscal y oferta de dólares; el sol tiende a apreciarse y la bolsa a subir. Para el inversionista extranjero ambos efectos se suman.
