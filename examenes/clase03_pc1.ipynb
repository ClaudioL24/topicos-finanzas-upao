## Parte 2: encuentra los errores (6 pts)

Un practicante valoró a la empresa Andina SAC y entregó este trabajo. Sus datos de partida son correctos: EBIT = 50, t = 30 por ciento, depreciación = 10, capex = 15, incremento del capital de trabajo = 5, gasto por intereses = 8, endeudamiento neto = 0, Ke = 12 por ciento, WACC = 9 por ciento.

> **Paso 1.** FCFF = EBIT(1-t) + Dep + Int - Capex - WCInv = 35 + 10 + 8 - 15 - 5 = **33**
>
> **Paso 2.** FCFE = FCFF - Int = 33 - 8 = **25**
>
> **Paso 3.** Como la empresa creció 7 por ciento en los últimos años, uso g = 7 por ciento a perpetuidad. Valor del equity = FCFE por (1+g) entre (WACC - g) = 25 por 1.07 entre 0.02 = **1,337.5**

El trabajo contiene **cuatro errores conceptuales**. Identifícalos (2 pts), explica en una oración por qué cada uno es un error (2 pts), y recalcula el valor del equity corrigiendo todo, con un g perpetuo de 3 por ciento (2 pts). Puedes usar la celda de código.

*Errores y explicaciones:*

1. En el paso 1: Esta partiendo del EBIT y esta aumentandole los interesés cuando estos no han son descontados del EBIT en primer lugar. 
2. En el paso 2: Esta restandole los intereses antes de los impuestos, se deberia tomar en cuenta el escudo fiscal que estos generan.
3. En el paso 3: Asumir que crece a 7% a perpetuidad es incorrecto, ningun crecimiento puede superar al de la economía de un país a perpetuidad, teniendo en cuenta que el crecimiento aproximado de un país esta en un ~4%; además de que el crecimiento histórico no representa los hechos futuros. 
4. En el paso 3: Esta utilizando el WACC para descontar el FCFE, incurriendo en la regla de consistencia.

````python
# Recalculo corregido (puedes usar utils/finanzas.py si trabajas en tu fork)

def fcff_desde_ebit(ebit: float, t: float, dep: float, fcinv: float, wcinv: float) -> float:
    """FCFF = EBIT(1-t) + Dep - FCInv - WCInv."""
    return ebit * (1 - t) + dep - fcinv - wcinv

fcff = fcff_desde_ebit(ebit=50, t=0.3, dep=10, fcinv=15, wcinv=5)
print(f"FCFF = {fcff:.2f}")

def valor_crecimiento_constante(cf1: float, r: float, g: float) -> float:
    """Valor presente de una perpetuidad creciente: V0 = CF1 / (r - g). Requiere g < r."""
    if g >= r:
        raise ValueError("El crecimiento g debe ser menor que la tasa de descuento r.")
    return cf1 / (r - g)

VE=valor_crecimiento_constante(cf1=fcff, r=0.09, g=0.03)
print(f"Valor de la empresa (VE) = {VE:.2f}")
````
